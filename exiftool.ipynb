```python
import os
import subprocess
import pandas as pd
import io

# 🔹 Ścieżka do folderu ze zdjęciami
#folders_list = ["E:/Obrazy/Japonia 2025/01_Tokyo/","E:/Obrazy/Japonia 2025/02_Okinawa/","E:/Obrazy/Japonia 2025/03_Hiroshima/","E:/Obrazy/Japonia 2025/04_Osaka/","E:/Obrazy/Japonia 2025/05_Kyoto/","E:/Obrazy/Japonia 2025/06_Tokyo/"]
#folder_path = r"E:/Obrazy/Japonia 2025/01_Tokyo/"
#folder_path = r"E:/Obrazy/Japonia 2025/02_Okinawa/"
#folder_path = r"E:/Obrazy/Japonia 2025/03_Hiroshima/"
#folder_path = r"E:/Obrazy/Japonia 2025/04_Osaka/"
#folder_path = r"E:/Obrazy/Japonia 2025/05_Kyoto/"
#folder_path = r"E:/Obrazy/Japonia 2025/06_Tokyo/"
#folder_path = r"E:/Obrazy/Japonia 2025/wybrane/"
folder_path = r"C:/Users/pitus/Pictures/Zieleniec 2025"

# Lista pól EXIF, które chcemy pobrać
fields = [
    "-FileName",
    "-LensID",
    "-FocalLength",
    "-ExposureTime",
    "-ISO",
    "-FNumber",
    "-MaxApertureValue",
    "-LightValue",  # EV
    "-ExposureProgram",
    "-HyperfocalDistance"   # Odległość ostrości
]

# Przygotowanie komendy exiftool
cmd = ["exiftool", "-csv"] + fields + ["-ext", "jpg", "-ext", "jpeg", folder_path]

# Uruchomienie exiftool
result = subprocess.run(cmd, capture_output=True, text=True)

# Wczytanie wyniku do DataFrame
df = pd.read_csv(io.StringIO(result.stdout))

# Wyświetlenie wyników
print(df)

# Zapis do pliku CSV
output_file = os.path.join(folder_path, "exif_output.csv")
df.to_csv(output_file, index=False)
print(f"Dane zapisane do: {output_file}")
```
